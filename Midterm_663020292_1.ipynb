### Task เขียนโคพเพื่ออ่านไฟล์ใน input ในแต่ละวัน แล้วเก็บเข้าตัวแปรที่มีชื่อเหมือนชื่อไฟล์เลย
Load daily rain data from the directory "/content/drive/MyDrive/BSC DPDM midterm 2025/Input - (Operation Daily rain from API_ 2025)/" by identifying files matching the 'YYYY-MM-DD.csv' pattern, and then dynamically load each file into a separate pandas DataFrame named after its date (e.g., '2025-01-01.csv' becomes a DataFrame named `_2025_01_01`). Finally, display the head of a few of these newly created daily DataFrames to verify the loading process.

```python
import os

input_folder_path = '/content/drive/MyDrive/BSC DPDM midterm 2025/Input - (Operation Daily rain from API_ 2025)/'
all_files = os.listdir(input_folder_path)
print(f"All files in the directory: {all_files}")
```

```python
import re

daily_file_pattern = re.compile(r'^\d{4}-\d{2}-\d{2}\.csv$')
daily_data_files = [f for f in all_files if daily_file_pattern.match(f)]

print(f"Daily data files identified: {daily_data_files}")

if not daily_data_files:
    print("No files matching 'YYYY-MM-DD.csv' pattern found directly in the input directory.")
```

```python
daily_data_files_found_in_subfolders = []

for item in all_files:
    item_path = os.path.join(input_folder_path, item)
    if os.path.isdir(item_path): # Check if it's a directory
        print(f"Searching in subdirectory: {item_path}")
        sub_files = os.listdir(item_path)
        for f in sub_files:
            if daily_file_pattern.match(f):
                daily_data_files_found_in_subfolders.append(os.path.join(item_path, f))

print(f"Daily data files identified in subfolders: {daily_data_files_found_in_subfolders}")

if not daily_data_files_found_in_subfolders:
    print("No files matching 'YYYY-MM-DD.csv' pattern found in any subdirectories.")
```

```python
daily_file_pattern = re.compile(r'.*\d{4}-\d{2}-\d{2}\.csv$') # Updated regex to match files with a prefix
daily_data_files_found_in_subfolders = []

for item in all_files:
    item_path = os.path.join(input_folder_path, item)
    if os.path.isdir(item_path): # Check if it's a directory
        print(f"Searching in subdirectory: {item_path}")
        sub_files = os.listdir(item_path)
        for f in sub_files:
            if daily_file_pattern.match(f):
                daily_data_files_found_in_subfolders.append(os.path.join(item_path, f))

print(f"Daily data files identified in subfolders: {daily_data_files_found_in_subfolders}")

if not daily_data_files_found_in_subfolders:
    print("No files matching 'YYYY-MM-DD.csv' pattern found in any subdirectories with the updated regex.")
```

```python
daily_dataframes = {}
print("Initialized empty dictionary 'daily_dataframes' to store loaded DataFrames.")
```

```python
import re
import pandas as pd
import os

for file_path in daily_data_files_found_in_subfolders:
    try:
        # Extract filename from the full path
        filename = os.path.basename(file_path)

        # Extract the date string (e.g., 'YYYY-MM-DD') from filenames like 'HII_daily_rainfall_YYYY-MM-DD.csv'
        match = re.search(r'\d{4}-\d{2}-\d{2}', filename)
        if match:
            date_str = match.group(0)
            # Create a valid variable name (e.g., '_2025_01_01')
            var_name = '_' + date_str.replace('-', '_')

            # Read the CSV file into a pandas DataFrame
            df = pd.read_csv(file_path)

            # Store the loaded DataFrame in the daily_dataframes dictionary
            daily_dataframes[var_name] = df
            print(f"Successfully loaded '{filename}' as '{var_name}'.")
        else:
            print(f"Skipping file '{filename}': Could not extract date in YYYY-MM-DD format.")

    except FileNotFoundError:
        print(f"Error: File not found at '{file_path}'.")
    except pd.errors.EmptyDataError:
        print(f"Error: '{file_path}' is empty and cannot be parsed.")
    except Exception as e:
        print(f"Error processing '{file_path}': {e}")

# Display the head of a few newly created daily DataFrames to verify the loading process
print("\nDisplaying head of a few loaded DataFrames (if available):")
count = 0
for var_name, df in list(daily_dataframes.items())[:3]: # Display first 3 loaded DataFrames
    print(f"\nDataFrame: {var_name}")
    print(df.head())
    count += 1
if count == 0:
    print("No daily DataFrames were loaded.")
```

## สัดส่วนข้อมูลที่หายไปในแต่ละสถานี

```python
import pandas as pd

result = []

stations = all_daily_df["tele_station_oldcode"].unique()

for station in stations:
    total = len(all_daily_df[all_daily_df["tele_station_oldcode"] == station])
    remain = len(filtered_df[filtered_df["tele_station_oldcode"] == station])

    if total > 0:
        missing_ratio = (total - remain) / total
    else:
        missing_ratio = 0

    result.append({
        "station": station,
        "total_data": total,
        "remain_data": remain,
        "missing_data": total - remain,
        "missing_ratio": missing_ratio
    })

missing_summary = pd.DataFrame(result)

missing_summary
```

## สร้าง day of year จากคอลัมน์ measure_datetime

```python
df = filtered_df.copy()

df["measure_datetime"] = pd.to_datetime(df["measure_datetime"])
df["day_of_year"] = df["measure_datetime"].dt.dayofyear
```

```python
df
```

## เติมค่า median ตาม day of year แยกตาม สถานี

```python
# 1) median ตาม สถานี + day of year
median_station_doy = (
    df.groupby(["tele_station_oldcode", "day_of_year"])["rainfall_value"]
      .median()
      .reset_index()
      .rename(columns={"rainfall_value": "median_doy"})
)

df = df.merge(
    median_station_doy,
    on=["tele_station_oldcode", "day_of_year"],
    how="left"
)

df["rainfall_value"] = df["rainfall_value"].fillna(df["median_doy"])

# 2) median ทั้งสถานี (fallback)
median_station = df.groupby("tele_station_oldcode")["rainfall_value"].median()

df["rainfall_value"] = df["rainfall_value"].fillna(
    df["tele_station_oldcode"].map(median_station)
)

# ---------- เติม tele_station_id ----------

# หา tele_station_id ที่พบบ่อยที่สุดของแต่ละสถานี
station_id_mode = (
    df.groupby("tele_station_oldcode")["tele_station_id"]
      .agg(lambda x: x.dropna().mode().iloc[0] if not x.dropna().empty else None)
)

df["tele_station_id"] = df["tele_station_id"].fillna(
    df["tele_station_oldcode"].map(station_id_mode)
)

# ---------- ล้างคอลัมน์ช่วย ----------
df = df.drop(columns=["median_doy"])
```
