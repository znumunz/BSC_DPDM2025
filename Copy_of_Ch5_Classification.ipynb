# Task
The overall task is to perform a stratified train-test split on the `digits` dataset, then use GridSearchCV with 10-fold cross-validation to find the best hyperparameters and evaluate the performance of both K-Nearest Neighbors and Decision Tree classifiers, and finally select and evaluate the best performing model on the held-out test set.

## Stratified Train-Test Split

### Subtask:
Perform a stratified split of the dataset into training and testing sets, ensuring both sets have the same class distribution.


**Reasoning**:
The subtask requires performing a stratified train-test split. I need to import the necessary function from `sklearn.model_selection` and then apply it to the dataset with the specified parameters to create the training and testing sets.



```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)

print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

## Define Parameter Grids

### Subtask:
Define hyperparameter grids for both K-Nearest Neighbors (KNN) and Decision Tree classifiers for the Grid Search.


**Reasoning**:
To define the hyperparameter grids for K-Nearest Neighbors and Decision Tree classifiers, I need to import the respective classifier classes and then create dictionaries with the specified hyperparameters and their potential values.



```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

knn_param_grid = {
    'n_neighbors': [3, 5, 7, 9],
    'weights': ['uniform', 'distance']
}

dt_param_grid = {
    'max_depth': [None, 10, 20, 30],
    'min_samples_leaf': [1, 5, 10],
    'criterion': ['gini', 'entropy']
}

print("KNN Parameter Grid:", knn_param_grid)
print("Decision Tree Parameter Grid:", dt_param_grid)
```

## Configure and Run GridSearchCV for KNN

### Subtask:
Set up and execute GridSearchCV for the K-Nearest Neighbors classifier using 10-fold cross-validation and accuracy as the scoring metric.


**Reasoning**:
To configure and run GridSearchCV for KNN, I need to import the necessary class, create a KNN classifier, instantiate GridSearchCV with the defined parameters, fit it to the training data, and then print the best parameters and best score as per the instructions.



```python
from sklearn.model_selection import GridSearchCV

knn = KNeighborsClassifier()

grid_search_knn = GridSearchCV(estimator=knn, param_grid=knn_param_grid, cv=10, scoring='accuracy', n_jobs=-1)

grid_search_knn.fit(X_train, y_train)

print("Best parameters for KNN:", grid_search_knn.best_params_)
print("Best accuracy score for KNN:", grid_search_knn.best_score_)
```

## Configure and Run GridSearchCV for Decision Tree

### Subtask:
Set up and execute GridSearchCV for the Decision Tree classifier using 10-fold cross-validation and accuracy as the scoring metric.


**Reasoning**:
To configure and run GridSearchCV for the Decision Tree, I need to instantiate a Decision Tree classifier, then create a GridSearchCV object with the defined parameter grid and cross-validation settings, fit it to the training data, and finally print the best parameters and score as requested in the subtask.



```python
dt = DecisionTreeClassifier()

grid_search_dt = GridSearchCV(estimator=dt, param_grid=dt_param_grid, cv=10, scoring='accuracy', n_jobs=-1)

grid_search_dt.fit(X_train, y_train)

print("Best parameters for Decision Tree:", grid_search_dt.best_params_)
print("Best accuracy score for Decision Tree:", grid_search_dt.best_score_)
```
